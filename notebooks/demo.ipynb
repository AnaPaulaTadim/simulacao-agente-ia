# Agente inteligente para controle de combustível

Demonstração interativa do projeto. Execute as células em ordem.

## 1. Configuração

```python
import sys
sys.path.insert(0, '../src')

%matplotlib inline
from combustivel import (
    Ambiente, AgenteReativo, Config, comparar, passo_a_passo, resumir, simular,
)
from combustivel.terminal import formatar_evento, formatar_resumo
from combustivel.visualizacao import grafico_simulacao, grafico_comparacao

config = Config(capacidade_l=50, semente=42)
config
```

## 2. Acompanhando dia a dia

`passo_a_passo` é um gerador: cada dia entrega a percepção, a decisão do agente (com o motivo) e o custo acumulado.

```python
DIAS = 60
ambiente = Ambiente(config)
agente = AgenteReativo(config)
eventos = []

for ev in passo_a_passo(agente, ambiente, DIAS):
    eventos.append(ev)
    print(formatar_evento(ev, DIAS))

resultado = resumir(agente.nome, ambiente)
print()
print(formatar_resumo(resultado, DIAS, eventos))
```

## 3. Gráfico final da simulação

```python
grafico_simulacao(ambiente)
```

## 4. Comparação entre estratégias (500 simulações)

```python
medias = comparar(config, dias=60, simulacoes=500)
for nome, m in medias.items():
    print(f"{nome:<32} R$/km = {m['custo_por_km']:.4f} | R$/L = {m['preco_medio_pago']:.3f}")
grafico_comparacao(medias)
```

## 5. Experimente

Mude `nivel_minimo`, `desconto` ou `janela` em `AgenteReativo` e veja o efeito no custo.

```python
for desconto in (1.00, 0.97, 0.95, 0.90, 0.85):
    custos = []
    for i in range(200):
        cfg = Config(semente=1000 + i)
        r, _ = simular(AgenteReativo(cfg, desconto=desconto), cfg, 60)
        custos.append(r.custo_por_km)
    print(f'desconto={desconto:.2f}  custo médio por km = R$ {sum(custos)/len(custos):.4f}')
```
